```sql
create or replace view workspace.dimension.dim_time as
SELECT 
  explode(sequence(1, 86400)) AS TIME_ID
, FLOOR((TIME_ID - 1) / 3600) AS HOUR
, FLOOR(((TIME_ID - 1) % 3600) / 60) AS MINUTE
, (TIME_ID - 1) % 60 AS SECOND
, LPAD(FLOOR((TIME_ID - 1) / 3600), 2, '0') || ':' || 
  LPAD(FLOOR(((TIME_ID - 1) % 3600) / 60), 2, '0') || ':' || 
  LPAD((TIME_ID - 1) % 60, 2, '0') AS TIME_24HR
, LPAD(
    CASE 
      WHEN FLOOR((TIME_ID - 1) / 3600) = 0 THEN 12
      WHEN FLOOR((TIME_ID - 1) / 3600) > 12 THEN FLOOR((TIME_ID - 1) / 3600) - 12
      ELSE FLOOR((TIME_ID - 1) / 3600)
    END, 2, '0') || ':' || 
  LPAD(FLOOR(((TIME_ID - 1) % 3600) / 60), 2, '0') || ':' || 
  LPAD((TIME_ID - 1) % 60, 2, '0') || ' ' ||
  CASE WHEN FLOOR((TIME_ID - 1) / 3600) < 12 THEN 'AM' ELSE 'PM' END AS TIME_12HR
, CASE 
    WHEN FLOOR((TIME_ID - 1) / 3600) >= 9 AND FLOOR((TIME_ID - 1) / 3600) < 17 
    THEN TRUE 
    ELSE FALSE 
  END AS IS_BUSINESS_HOURS
```
