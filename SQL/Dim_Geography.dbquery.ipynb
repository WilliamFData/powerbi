```sql
create or replace view workspace.dimension.dim_geography as 
select 
    MB_CODE21 as MB_CODE
  , MB_CAT21 as MB_CAT
  , SA1_CODE21 as SA1_CODE
  , SA2_CODE21 as SA2_CODE
  , SA2_NAME21 as SA2_NAME
  , SA3_CODE21 as SA3_CODE
  , SA3_NAME21 as SA3_NAME
  , SA4_CODE21 as SA4_CODE
  , SA4_NAME21 as SA4_NAME
  , GCC_CODE21 as GCC_CODE
  , GCC_NAME21 as GCC_NAME
  , STE_CODE21 as STE_CODE
  , STE_NAME21 as STE_NAME
  , AUS_CODE21 as AUS_CODE
  , AUS_NAME21 as AUS_NAME
  , AREASQKM21 as AREASQKM
  , lga.LGA_CODE_2025 as LGA_CODE
  , lga.LGA_NAME_2025 as LGA_NAME
  , sed.SED_CODE_2025 as SED_CODE
  , sed.SED_NAME_2025 as SED_NAME
from workspace.reference.mesh_blocks mb
  left join workspace.reference.abs_lga_data lga
    on mb.MB_CODE21 = lga.MB_CODE_2021
  left join workspace.reference.abs_sed_data sed
    on mb.MB_CODE21 = sed.MB_CODE_2021
where 1=1
```
